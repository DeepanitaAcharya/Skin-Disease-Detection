```python
import pandas as pd

fitz_path = "/content/drive/MyDrive/Skin_Disease_Detection/fitzpatrick17k.csv"

fitz = pd.read_csv(fitz_path)

print("Number of images:", len(fitz))
print("\nColumns:")
print(fitz.columns.tolist())
```

```python
print("Number of unique disease classes:")
print(fitz["label"].nunique())

print("\nDisease classes:")
print(fitz["label"].value_counts())
```

```python
target_classes = [
      "acne vulgaris",
      "eczema",
      "psoriasis",
      "rosacea",
      "melasma"
     ]

for disease in target_classes:
   matches = fitz[fitz["label"].str.lower().str.contains(disease, na=False)]

print(f"\n{disease}")
print("Images:", len(matches))
print("Labels found:", matches["label"].unique())
```

```python
target_labels = [
      "acne",
      "acne vulgaris",
      "eczema",
      "dyshidrotic eczema",
      "psoriasis",
      "pustular psoriasis",
      "rosacea",
      "lichen planus"
   ]

filtered_fitz = fitz[fitz["label"].isin(target_labels)].copy()

print("Total selected images:", len(filtered_fitz))
print("\nSelected labels:")
print(filtered_fitz["label"].value_counts())
```

```python
filtered_fitz["disease"] = filtered_fitz["label"].replace({
      "acne": "Acne",
      "acne vulgaris": "Acne",
      "eczema": "Eczema",
      "dyshidrotic eczema": "Eczema",
      "psoriasis": "Psoriasis",
      "pustular psoriasis": "Psoriasis",
      "rosacea": "Rosacea",
      "lichen planus": "Lichen planus"
})

print(filtered_fitz["disease"].value_counts())
```

```python
#testing only 5 images

import requests

test_urls = filtered_fitz["url"].head(5).tolist()

for i, url in enumerate(test_urls, 1):
   try:
        response = requests.get(url, timeout=10)
        print(f"Image {i}: Status {response.status_code}, Size {len(response.content)} bytes")
   except Exception as e:
        print(f"Image {i}: Failed - {e}")
```
